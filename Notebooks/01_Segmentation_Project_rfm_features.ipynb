```python
import pandas as pd
import numpy as np
import os
```

```python
os.makedirs('/content/drive/MyDrive/segmentation_project', exist_ok=True)
```

```python
churn_folder = ('/content/drive/MyDrive/churn_project/')
df = pd.read_csv(churn_folder + 'purchase_clean.csv', parse_dates = ['InvoiceDate'],  dtype={'Invoice': str, 'StockCode': str})
```

```python
df.shape
```

```python
df.head()
```

```python
snapshot_date = df['InvoiceDate'].max() + pd.Timedelta(days=1)

print('Last purchase in data:', df['InvoiceDate'].max())
print('Snapshot:', snapshot_date)
```

```python
customer_groups = df.groupby('Customer ID')
```

```python
last_purchase = customer_groups['InvoiceDate'].max()
recency = (snapshot_date - last_purchase).dt.days
frequency = customer_groups['Invoice'].nunique()
monetary = customer_groups['Revenue'].sum()
```

```python
rfm = pd.DataFrame({'Recency': recency, 'Frequency': frequency, 'Monetary': monetary})
rfm.index.name = 'Customer ID'
rfm = rfm.reset_index()

print('Customers:', len(rfm))
rfm.head()
```

```python
print('Missing values:', rfm.isnull().sum().sum())
```

```python
rfm.drop('Customer ID', axis=1).describe().round(1)
```

```python
rfm.to_csv('/content/drive/MyDrive/segmentation_project/rfm.csv', index=False)
```
